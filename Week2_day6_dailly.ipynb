```python
#Step 1: Create the Farm Class
class Farm:
    def __init__(self, farm_name):
        self.name = farm_name
        self.animals = {}

    def add_animal(self, animal=None, count=1, **kwargs):
        # Handle keyword arguments like add_animal(cow=5, sheep=2)
        if kwargs:
            for animal_name, qty in kwargs.items():
                if animal_name in self.animals:
                    self.animals[animal_name] += qty
                else:
                    self.animals[animal_name] = qty
        # Handle standard positional call like add_animal("cow", 5)
        elif animal is not None:
            if animal in self.animals:
                self.animals[animal] += count
            else:
                self.animals[animal] = count

    def get_info(self):
        info = f"{self.name}'s farm\n"
        for animal, count in self.animals.items():
            info += f"{animal:<10} : {count}\n"
        info += "E-I-E-I-O!"
        return info

    def get_animal_types(self):
        return sorted(list(self.animals.keys()))

    def get_short_info(self):
        animal_types = self.get_animal_types()
        animals = []
        for animal in animal_types:
            if self.animals[animal] > 1:
                animals.append(animal + "s")
            else:
                animals.append(animal)

        if len(animals) == 1:
            animal_list = animals[0]
        elif len(animals) == 2:
            animal_list = f"{animals[0]} and {animals[1]}"
        else:
            animal_list = ", ".join(animals[:-1]) + " and " + animals[-1]

        return f"{self.name}'s farm has {animal_list}."


# Create a farm
Apolo = Farm("Apolo")
#Step 2: Implement the __init__ Method
# Add animals
Apolo.add_animal("cow", 5)
Apolo.add_animal("sheep", 10)
Apolo.add_animal("goat", 3)
Apolo.add_animal("cow", 2)

# Display farm information
print(Apolo.get_info())
```
